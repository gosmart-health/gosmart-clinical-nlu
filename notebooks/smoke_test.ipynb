# Smoke test — `gosmart_clinical_nlu`

Quick environment sanity check for the **gosmart-clinical-nlu** package and its
runtime dependencies (`spacy`, `pydantic`, `dateparser`).

Run all cells top-to-bottom. Any traceback means the environment is broken.

```python
import sys

import gosmart_clinical_nlu

print("python:", sys.version.split()[0])
print("gosmart_clinical_nlu:", gosmart_clinical_nlu.__version__)
print("imported from:", gosmart_clinical_nlu.__file__)

assert gosmart_clinical_nlu.__version__ == "0.1.0"
print("OK: package imports and version matches")
```

```python
from gosmart_clinical_nlu import print_hi

print_hi("smoke test")
print("OK: public API callable")
```

```python
import spacy

nlp = spacy.blank("en")  # blank pipeline: no model download needed
doc = nlp("Patient admitted on 3 March 2025 with chest pain.")
print("spacy:", spacy.__version__)
print("tokens:", [t.text for t in doc])
print("OK: spacy pipeline runs")
```

```python
from datetime import date

import pydantic
from pydantic import BaseModel


class Admission(BaseModel):
    patient_id: str
    admitted_on: date


admission = Admission(patient_id="PT-001", admitted_on="2025-03-03")
print("pydantic:", pydantic.__version__)
print(admission)
assert admission.admitted_on == date(2025, 3, 3)
print("OK: pydantic validation works")
```

```python
import dateparser
from dateparser.search import search_dates

text = "patient admitted on 3 March 2025 and discharged 2 weeks later"
print("dateparser:", dateparser.__version__)
print("parse:", dateparser.parse("3 March 2025"))
print("search:", search_dates(text))
print("OK: dateparser parses and searches")
```

## Result

If every cell above printed its output without a traceback, the package and
all three runtime dependencies are healthy in this environment.
